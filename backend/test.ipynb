```python
from docAI import ChatInterface  
from IPython.display import display, Markdown
```

```python
chat = ChatInterface()

try:
    chunks = chat.loadPdf("./data/document/Lecture 8 Postclass Software Engineering.pdf")        
    chat.storeChunks()
except Exception as e:
    print(f"An error occurred: {str(e)}")
    print("Make sure the PDF file exists and the path is correct.")
    print("Also ensure that all required dependencies are installed.")
```

```python
has_docs = chat.checkDocument()

query = input("Input your question here: ")
if has_docs:
    res = chat.chat(query)
    display(Markdown(res))
else:
    print("No documents were found in the vector store. Check if the PDF was loaded correctly.")
```
